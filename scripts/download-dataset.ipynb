# Download Dataset Olist — Brazilian E-Commerce

Notebook ini men-download dataset Olist dari Kaggle via `kagglehub`, memverifikasi kelengkapan semua file, lalu menyalinnya ke folder `data/raw/` di repo project.

**Prasyarat:** token Kaggle sudah tersimpan di `~/.kaggle/access_token` (lihat langkah setup sebelumnya).

```python
# Install kagglehub jika belum terpasang di environment
# Uncomment baris di bawah kalau perlu
# !pip install kagglehub
```

```python
import kagglehub
import shutil
from pathlib import Path

# Download dataset resmi Olist dari Kaggle
path = kagglehub.dataset_download("olistbr/brazilian-ecommerce")

print("Dataset berhasil di-download ke cache:")
print(path)
```

## Verifikasi Kelengkapan File

Pastikan ke-9 file CSV ada, termasuk `product_category_name_translation.csv` yang sempat hilang sebelumnya.

```python
expected_files = [
    "olist_customers_dataset.csv",
    "olist_geolocation_dataset.csv",
    "olist_order_items_dataset.csv",
    "olist_order_payments_dataset.csv",
    "olist_order_reviews_dataset.csv",
    "olist_orders_dataset.csv",
    "olist_products_dataset.csv",
    "olist_sellers_dataset.csv",
    "product_category_name_translation.csv",
]

cache_dir = Path(path)
found_files = [f.name for f in cache_dir.glob("*.csv")]

missing = [f for f in expected_files if f not in found_files]

print(f"Total file CSV ditemukan: {len(found_files)}")
if missing:
    print("File yang HILANG:")
    for f in missing:
        print(f"   - {f}")
else:
    print("Semua 9 file lengkap.")
```

## Salin ke Struktur Repo

Copy (bukan move) semua CSV ke `data/raw/` di repo project, supaya cache di kagglehub tetap jadi backup.

```python
# Notebook ini disarankan ada di notebooks/00_download_data.ipynb,
# jadi data/raw/ ada satu level di atas folder notebooks/
repo_raw_dir = Path("../data/raw")

repo_raw_dir.mkdir(parents=True, exist_ok=True)

for filename in expected_files:
    src = cache_dir / filename
    dst = repo_raw_dir / filename
    if src.exists():
        shutil.copy(src, dst)
        print(f"Copied: {filename}")
    else:
        print(f"Dilewati (tidak ditemukan di cache): {filename}")

print(f"\nSemua file tersedia di: {repo_raw_dir.resolve()}")
```

## Selesai

Data mentah sudah ada di `data/raw/`. Pastikan `data/raw/*.csv` sudah masuk `.gitignore` supaya tidak ikut ter-push ke GitHub — cukup notebook ini yang di-commit untuk reproducibility.
